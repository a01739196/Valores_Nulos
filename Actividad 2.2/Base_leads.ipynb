```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel('Base_de_Leads.xlsx')
data.head(5)
```

```python
#Identificar valores nulos por columna
valores_nulos = data.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos en todo el dataframe
valores_nulos = data.isnull().sum().sum()
valores_nulos
```

```python
## Eliminar columnas que no aportan
#Las columnas Telefono, correo no tienen ningun contenido
data1 = data.drop(['Telefono',
                    'Correo'
                   ], axis=1)
data1
```

```python
#Corroboramos valores nulos
valores_nulos = data1.isnull().sum()
valores_nulos
```

```python
#Renombramos la columna sin encabezado (contiene el estatus del lead: Ventas, Finalizado, Contactado)
data1 = data1.rename(columns={' ': 'Estatus'})
data1
```

```python
##Metodos de sustitución de valores nulos
#Realizamos una copia del dataframe
data2 = data1.copy()
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto (columna categorica con pocos nulos)
data2["Producto"] = data2["Producto"].fillna("No especificado")
```

```python
#Corroboramos valores nulos en todo el dataframe
valores_nulos = data2.isnull().sum().sum()
valores_nulos
```

```python
#Convertir DataFrame a CSV
data2.to_csv("Base_de_Leads_sin_nulos.csv", index=False)
```
